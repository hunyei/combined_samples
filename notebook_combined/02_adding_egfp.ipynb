```python
"egfp" in m5_adata.var_names
```

```python
with h5py.File('/blue/square.t/peter.huynh/cellranger/cellranger_outs/cellranger_minus345_2_v5e_062522_forced6k/outs/minus345_filtered_feature_bc_matrix.h5', 'r') as f:
    barcodes = [b.decode() for b in f['matrix']['barcodes'][:]]
    genes = [g.decode() for g in f['matrix']['features']['name'][:]]
    egfp_idx = genes.index("egfp")
    X = sparse.csc_matrix((f['matrix']['data'][:], f['matrix']['indices'][:], f['matrix']['indptr'][:]),
                          shape=f['matrix']['shape'][:])
    egfp_vector = X[egfp_idx].toarray().flatten()

# Align to AnnData barcodes
formatted_barcodes = ["minus345_2_v5e_062522_forced6k_possorted_genome_CEA3D:" + bc.replace("-1", "x") for bc in barcodes]
egfp_series = pd.Series(egfp_vector, index=formatted_barcodes)
aligned = egfp_series.reindex(m5_adata.obs_names).fillna(0).astype(int).values.reshape(-1, 1)

# Inject into AnnData
egfp_col = m5_adata.var_names.get_loc("egfp")
m5_adata.X = m5_adata.X.tolil()
m5_adata.X[:, egfp_col] = aligned
m5_adata.X = m5_adata.X.tocsr()
```
